```python
"""
Improved EEG Motor Imagery Classification
==========================================
Key improvements over original:
1. Proper frequency-domain feature extraction (band powers)
2. Intelligent 4-channel selection based on motor cortex regions
3. Temporal windowing with sliding windows
4. Proper train/test split with stratification
5. Advanced feature engineering (PSD, band ratios, statistical features)
6. Multiple model comparison
7. Cross-validation for robust evaluation
"""

import os
import time
import warnings
import numpy as np
import pandas as pd
from collections import Counter
from scipy import signal
from scipy.stats import skew, kurtosis
import mne
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.utils.class_weight import compute_class_weight
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, mutual_info_classif
import joblib

warnings.filterwarnings('ignore')

# ---------- CONFIG ----------
DEFAULT_DATASET_DIR = "/project/ece601/EEG ALS/EEGET-ALS Dataset"

# =============================================================================
# YOUR HEADSET'S 4 CHANNELS: Fp1, Fp2, T7, T8
# =============================================================================
# - Fp1, Fp2: Frontal pole (forehead) - captures frontal activity and 
#             some motor planning signals
# - T7, T8: Temporal (above ears) - can pick up motor activity via 
#           volume conduction from nearby motor cortex

# Priority list - your 4 channels first, then fallbacks if names differ in dataset
MOTOR_CHANNELS_PRIORITY = [
    # YOUR PRIMARY CHANNELS
    'Fp1', 'Fp2', 'T7', 'T8',
    # Alternative names that might appear in datasets
    'FP1', 'FP2',  # uppercase variant
    'T3', 'T4',    # old 10-20 naming (T3=T7, T4=T8)
    # Nearby alternatives if exact match not found
    'F7', 'F8',    # lateral frontal (close to Fp1/Fp2)
    'AF7', 'AF8',  # anterior frontal
    'FT7', 'FT8',  # frontal-temporal (close to T7/T8)
    'TP7', 'TP8',  # temporal-parietal
]

# Your exact 4 channels
IDEAL_4_CHANNELS = ['Fp1', 'Fp2', 'T7', 'T8']

# EEG frequency bands (Hz)
FREQ_BANDS = {
    'delta': (0.5, 4),
    'theta': (4, 8),
    'alpha': (8, 13),      # Important for motor imagery (ERD/ERS)
    'low_beta': (13, 20),  # Important for motor imagery
    'high_beta': (20, 30),
    'gamma': (30, 45),
}

# Scenario labels
SCENARIO_LABELS = {
    "scenario1": "left_hand",
    "scenario2": "right_hand", 
    "scenario3": "left_leg",
    "scenario4": "right_leg",
    "scenario5": "mouth",
    "scenario6": "nod_head",
    "scenario7": "shake_head"
}
IGNORE_SCENARIOS = {"scenario8", "scenario9"}
SEGMENT_TYPES = ["Thinking", "Resting", "Think_Acting", "Typing"]

# ----------------------------


class EEGFeatureExtractor:
    """
    Extract meaningful features from EEG signals for motor imagery classification.
    """
    
    def __init__(self, sfreq=256, n_channels=4, window_sec=2.0, overlap=0.5):
        self.sfreq = sfreq
        self.n_channels = n_channels
        self.window_samples = int(window_sec * sfreq)
        self.overlap_samples = int(self.window_samples * overlap)
        self.step = self.window_samples - self.overlap_samples
        
    def compute_band_powers(self, segment):
        """
        Compute power spectral density in each frequency band.
        This is THE MOST IMPORTANT feature for motor imagery!
        """
        band_powers = {}
        
        for band_name, (low, high) in FREQ_BANDS.items():
            # Use Welch's method for robust PSD estimation
            freqs, psd = signal.welch(segment, fs=self.sfreq, nperseg=min(256, len(segment)))
            
            # Find frequency band indices
            idx_band = np.logical_and(freqs >= low, freqs <= high)
            
            # Average power in band (log transform for better distribution)
            if np.any(idx_band):
                band_powers[band_name] = np.log10(np.mean(psd[idx_band]) + 1e-10)
            else:
                band_powers[band_name] = 0
                
        return band_powers
    
    def compute_statistical_features(self, segment):
        """
        Compute time-domain statistical features.
        """
        features = {
            'mean': np.mean(segment),
            'std': np.std(segment),
            'var': np.var(segment),
            'max': np.max(segment),
            'min': np.min(segment),
            'peak_to_peak': np.ptp(segment),
            'skewness': skew(segment),
            'kurtosis': kurtosis(segment),
            'rms': np.sqrt(np.mean(segment**2)),
            'zero_crossings': np.sum(np.diff(np.signbit(segment))),
        }
        return features
    
    def compute_hjorth_parameters(self, segment):
        """
        Hjorth parameters: Activity, Mobility, Complexity
        Classic EEG features that capture signal dynamics.
        """
        # Activity (variance)
        activity = np.var(segment)
        
        # First derivative
        diff1 = np.diff(segment)
        
        # Second derivative
        diff2 = np.diff(diff1)
        
        # Mobility
        mobility = np.sqrt(np.var(diff1) / (activity + 1e-10))
        
        # Complexity
        complexity = np.sqrt(np.var(diff2) / (np.var(diff1) + 1e-10)) / (mobility + 1e-10)
        
        return {
            'hjorth_activity': activity,
            'hjorth_mobility': mobility,
            'hjorth_complexity': complexity
        }
    
    def extract_features_single_channel(self, channel_data):
        """
        Extract all features from a single channel.
        """
        features = {}
        
        # Band powers (most important!)
        features.update(self.compute_band_powers(channel_data))
        
        # Statistical features
        features.update(self.compute_statistical_features(channel_data))
        
        # Hjorth parameters
        features.update(self.compute_hjorth_parameters(channel_data))
        
        # Band power ratios (important for motor imagery)
        if 'alpha' in features and 'low_beta' in features:
            features['alpha_beta_ratio'] = features['alpha'] / (features['low_beta'] + 1e-10)
        if 'theta' in features and 'alpha' in features:
            features['theta_alpha_ratio'] = features['theta'] / (features['alpha'] + 1e-10)
            
        return features
    
    def extract_features(self, segment_data, channel_names=None):
        """
        Extract features from multi-channel EEG segment.
        
        Args:
            segment_data: (n_samples, n_channels) array
            channel_names: list of channel names
            
        Returns:
            Feature vector (1D array)
        """
        n_samples, n_channels = segment_data.shape
        all_features = []
        feature_names = []
        
        for ch_idx in range(n_channels):
            ch_name = channel_names[ch_idx] if channel_names else f"ch{ch_idx}"
            ch_data = segment_data[:, ch_idx]
            
            # Skip if channel has invalid data
            if np.isnan(ch_data).any() or np.std(ch_data) < 1e-10:
                # Fill with zeros for this channel
                ch_features = self.extract_features_single_channel(np.random.randn(len(ch_data)) * 1e-6)
            else:
                ch_features = self.extract_features_single_channel(ch_data)
            
            for feat_name, feat_val in ch_features.items():
                all_features.append(feat_val)
                feature_names.append(f"{ch_name}_{feat_name}")
        
        # Inter-channel features (connectivity-like)
        if n_channels >= 2:
            for i in range(n_channels):
                for j in range(i+1, n_channels):
                    # Correlation between channels
                    corr = np.corrcoef(segment_data[:, i], segment_data[:, j])[0, 1]
                    if np.isnan(corr):
                        corr = 0
                    all_features.append(corr)
                    ch_i = channel_names[i] if channel_names else f"ch{i}"
                    ch_j = channel_names[j] if channel_names else f"ch{j}"
                    feature_names.append(f"corr_{ch_i}_{ch_j}")
        
        return np.array(all_features), feature_names


def select_best_channels(raw, n_channels=4, preferred_channels=None):
    """
    Select the best N channels for motor imagery classification.
    Prioritizes channels over motor cortex regions.
    """
    available_channels = raw.ch_names
    available_lower = [ch.lower() for ch in available_channels]
    
    selected = []
    selected_indices = []
    
    # First, try to match preferred channels
    if preferred_channels is None:
        preferred_channels = MOTOR_CHANNELS_PRIORITY
    
    for pref in preferred_channels:
        if len(selected) >= n_channels:
            break
        pref_lower = pref.lower()
        
        # Try exact match first
        for idx, ch in enumerate(available_channels):
            if ch.lower() == pref_lower and idx not in selected_indices:
                selected.append(ch)
                selected_indices.append(idx)
                break
        else:
            # Try partial match
            for idx, ch in enumerate(available_channels):
                if pref_lower in ch.lower() and idx not in selected_indices:
                    selected.append(ch)
                    selected_indices.append(idx)
                    break
    
    # If we still don't have enough, add remaining channels
    if len(selected) < n_channels:
        for idx, ch in enumerate(available_channels):
            if idx not in selected_indices:
                # Skip obviously non-EEG channels
                ch_lower = ch.lower()
                if any(skip in ch_lower for skip in ['eog', 'emg', 'ecg', 'stim', 'status', 'trigger']):
                    continue
                selected.append(ch)
                selected_indices.append(idx)
                if len(selected) >= n_channels:
                    break
    
    return selected, selected_indices


def bandpass_filter(data, sfreq, low=0.5, high=45):
    """
    Apply bandpass filter to remove noise and artifacts.
    """
    nyq = sfreq / 2
    low_norm = low / nyq
    high_norm = min(high / nyq, 0.99)
    
    try:
        b, a = signal.butter(4, [low_norm, high_norm], btype='band')
        filtered = signal.filtfilt(b, a, data, axis=0)
        return filtered
    except Exception:
        return data


def load_and_preprocess_eeg(path=DEFAULT_DATASET_DIR, n_channels=4, window_sec=2.0, overlap=0.5):
    """
    Load EEG data with proper preprocessing and feature extraction.
    """
    path = os.path.abspath(path)
    print(f"🔍 Scanning for EDF files under: {path}")
    
    edf_files = []
    for root, _, files in os.walk(path):
        for fname in files:
            if fname.lower().endswith(".edf"):
                edf_files.append(os.path.join(root, fname))
    
    if not edf_files:
        raise RuntimeError(f"No EDF files found under {path}")
    
    print(f"✅ Found {len(edf_files)} EDF files")
    
    all_features = []
    all_labels = []
    feature_names = None
    
    # Track channel selection across files
    channel_selection_counts = Counter()
    
    for file_idx, f in enumerate(edf_files):
        # Determine scenario from path
        parts = [p.lower() for p in f.split(os.sep)]
        scenario = next((p for p in parts if "scenario" in p), None)
        if not scenario or scenario in IGNORE_SCENARIOS:
            continue
        
        scenario_label = SCENARIO_LABELS.get(scenario, scenario)
        
        try:
            # Load EDF
            raw = mne.io.read_raw_edf(f, preload=True, verbose=False)
            sfreq = raw.info.get('sfreq', 256.0)
            
            # Select best channels for motor imagery
            selected_channels, selected_indices = select_best_channels(raw, n_channels)
            
            if len(selected_channels) < n_channels:
                print(f"⚠️ {os.path.basename(f)}: Only {len(selected_channels)} channels available")
                continue
            
            for ch in selected_channels:
                channel_selection_counts[ch] += 1
            
            # Get data for selected channels only
            data = raw.get_data()[selected_indices, :].T  # (n_samples, n_channels)
            
            # Apply bandpass filter
            data = bandpass_filter(data, sfreq, low=0.5, high=45)
            
            # Check for valid data
            if np.isnan(data).any() or data.shape[0] < int(2.0 * sfreq):
                print(f"⚠️ Skipping {os.path.basename(f)}: Invalid data")
                continue
            
            # Initialize feature extractor
            extractor = EEGFeatureExtractor(
                sfreq=sfreq, 
                n_channels=n_channels,
                window_sec=window_sec,
                overlap=overlap
            )
            
            # Determine segment label from annotations or filename
            segment_type = None
            fname_lower = os.path.basename(f).lower()
            
            # Check annotations
            if hasattr(raw, "annotations") and len(raw.annotations) > 0:
                for desc in raw.annotations.description:
                    desc_l = str(desc).lower()
                    matched = next((s for s in SEGMENT_TYPES if s.lower() in desc_l), None)
                    if matched:
                        segment_type = matched.lower()
                        break
            
            # Fallback to filename
            if segment_type is None:
                matched = next((s for s in SEGMENT_TYPES if s.lower() in fname_lower), None)
                if matched:
                    segment_type = matched.lower()
            
            # Create full label
            if segment_type:
                full_label = f"{scenario_label}_{segment_type}"
            else:
                full_label = scenario_label
            
            # Extract features using sliding windows
            window_samples = int(window_sec * sfreq)
            step_samples = int(window_samples * (1 - overlap))
            
            n_windows = (data.shape[0] - window_samples) // step_samples + 1
            
            for win_idx in range(n_windows):
                start = win_idx * step_samples
                end = start + window_samples
                
                if end > data.shape[0]:
                    break
                
                segment = data[start:end, :]
                
                # Extract features
                features, feat_names = extractor.extract_features(segment, selected_channels)
                
                if feature_names is None:
                    feature_names = feat_names
                
                # Skip if features contain NaN
                if np.isnan(features).any():
                    continue
                
                all_features.append(features)
                all_labels.append(full_label)
            
            if file_idx % 20 == 0:
                print(f"📄 Processed {file_idx+1}/{len(edf_files)} files | Current samples: {len(all_features)}")
                
        except Exception as e:
            print(f"⚠️ Error processing {f}: {e}")
            continue
    
    if not all_features:
        raise RuntimeError("❌ No valid EEG segments were extracted!")
    
    # Convert to arrays
    X = np.array(all_features)
    y = np.array(all_labels)
    
    print(f"\n📊 Dataset Summary:")
    print(f"  Total samples: {len(X)}")
    print(f"  Features per sample: {X.shape[1]}")
    print(f"  Unique classes: {len(np.unique(y))}")
    
    print(f"\n🔌 Most frequently used channels:")
    for ch, count in channel_selection_counts.most_common(10):
        print(f"    {ch}: {count} files")
    
    print(f"\n🏷️ Class distribution:")
    for label, count in Counter(y).most_common():
        print(f"    {label}: {count}")
    
    return X, y, feature_names


def prepare_dataset(dataset_path=DEFAULT_DATASET_DIR, n_channels=4, 
                    test_size=0.2, min_samples=5, simplify_labels=True):
    """
    Prepare dataset with proper train/test split and class filtering.
    """
    print("=" * 60)
    print("LOADING AND PREPROCESSING EEG DATA")
    print("=" * 60)
    
    X, y, feature_names = load_and_preprocess_eeg(
        dataset_path, 
        n_channels=n_channels,
        window_sec=2.0,
        overlap=0.5
    )
    
    # Optionally simplify labels (remove segment type, keep only movement)
    if simplify_labels:
        print("\n🔄 Simplifying labels to movement types only...")
        y_simplified = []
        for label in y:
            # Extract just the movement part
            parts = label.split('_')
            if len(parts) >= 2:
                # Keep movement type (first 2 parts usually)
                movement = '_'.join(parts[:2]) if parts[0] in ['left', 'right'] else parts[0]
            else:
                movement = label
            y_simplified.append(movement)
        y = np.array(y_simplified)
        
        print(f"🏷️ Simplified class distribution:")
        for label, count in Counter(y).most_common():
            print(f"    {label}: {count}")
    
    # Filter classes with too few samples
    class_counts = Counter(y)
    valid_classes = [cls for cls, count in class_counts.items() if count >= min_samples]
    
    if len(valid_classes) < len(class_counts):
        print(f"\n⚠️ Removing classes with < {min_samples} samples:")
        removed = [cls for cls in class_counts if cls not in valid_classes]
        for cls in removed:
            print(f"    {cls}: {class_counts[cls]} samples")
        
        mask = np.isin(y, valid_classes)
        X = X[mask]
        y = y[mask]
    
    if len(np.unique(y)) < 2:
        raise ValueError("❌ Less than 2 classes remain!")
    
    # Encode labels
    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)
    label_names = list(label_encoder.classes_)
    
    # Train/test split with stratification
    X_train, X_test, y_train, y_test = train_test_split(
        X, y_encoded, 
        test_size=test_size, 
        stratify=y_encoded,
        random_state=42
    )
    
    # Scale features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    print(f"\n✅ Dataset prepared:")
    print(f"    Training samples: {len(X_train)}")
    print(f"    Test samples: {len(X_test)}")
    print(f"    Number of classes: {len(label_names)}")
    print(f"    Features per sample: {X_train.shape[1]}")
    
    return (X_train_scaled, y_train), (X_test_scaled, y_test), scaler, label_names, feature_names


def train_and_evaluate_models(train_data, test_data, label_names, n_jobs=1):
    """
    Train multiple models and compare performance.
    """
    X_train, y_train = train_data
    X_test, y_test = test_data
    
    print("\n" + "=" * 60)
    print("TRAINING AND EVALUATING MODELS")
    print("=" * 60)
    
    # Handle class imbalance
    classes = np.unique(y_train)
    class_weights = compute_class_weight('balanced', classes=classes, y=y_train)
    class_weight_dict = {cls: w for cls, w in zip(classes, class_weights)}
    
    # Define models to compare
    models = {
        'MLP_Deep': MLPClassifier(
            hidden_layer_sizes=(512, 256, 128, 64),
            activation='relu',
            solver='adam',
            alpha=1e-3,
            learning_rate='adaptive',
            learning_rate_init=1e-3,
            max_iter=500,
            early_stopping=True,
            validation_fraction=0.15,
            n_iter_no_change=30,
            random_state=42,
            verbose=False
        ),
        'MLP_Wide': MLPClassifier(
            hidden_layer_sizes=(1024, 512),
            activation='relu',
            solver='adam',
            alpha=1e-4,
            learning_rate='adaptive',
            learning_rate_init=5e-4,
            max_iter=500,
            early_stopping=True,
            validation_fraction=0.15,
            n_iter_no_change=30,
            random_state=42,
            verbose=False
        ),
        'RandomForest': RandomForestClassifier(
            n_estimators=300,
            max_depth=20,
            min_samples_split=5,
            min_samples_leaf=2,
            class_weight='balanced',
            n_jobs=n_jobs,
            random_state=42
        ),
        'GradientBoosting': GradientBoostingClassifier(
            n_estimators=200,
            max_depth=6,
            learning_rate=0.1,
            subsample=0.8,
            random_state=42
        ),
        'SVM_RBF': SVC(
            kernel='rbf',
            C=10,
            gamma='scale',
            class_weight='balanced',
            random_state=42
        ),
    }
    
    results = {}
    best_model = None
    best_accuracy = 0
    
    for name, model in models.items():
        print(f"\n{'─' * 40}")
        print(f"🚀 Training {name}...")
        
        try:
            # Train
            model.fit(X_train, y_train)
            
            # Predict
            y_pred_train = model.predict(X_train)
            y_pred_test = model.predict(X_test)
            
            # Calculate metrics
            train_acc = accuracy_score(y_train, y_pred_train)
            test_acc = accuracy_score(y_test, y_pred_test)
            
            results[name] = {
                'train_accuracy': train_acc,
                'test_accuracy': test_acc,
                'model': model
            }
            
            print(f"  Training accuracy: {train_acc:.4f}")
            print(f"  Test accuracy: {test_acc:.4f}")
            
            if test_acc > best_accuracy:
                best_accuracy = test_acc
                best_model = (name, model)
                
        except Exception as e:
            print(f"  ❌ Error: {e}")
            continue
    
    # Print detailed results for best model
    if best_model:
        name, model = best_model
        print(f"\n{'=' * 60}")
        print(f"🏆 BEST MODEL: {name} (Test Accuracy: {best_accuracy:.4f})")
        print(f"{'=' * 60}")
        
        y_pred = model.predict(X_test)
        
        print("\n📊 Classification Report:")
        print(classification_report(y_test, y_pred, target_names=label_names, zero_division=0))
        
        print("\n📉 Confusion Matrix:")
        cm = confusion_matrix(y_test, y_pred)
        print(pd.DataFrame(cm, index=label_names, columns=label_names))
    
    return results, best_model


def cross_validate_best_model(X, y, model, cv=5):
    """
    Perform cross-validation for more robust evaluation.
    """
    print(f"\n{'─' * 40}")
    print(f"🔄 Performing {cv}-fold cross-validation...")
    
    skf = StratifiedKFold(n_splits=cv, shuffle=True, random_state=42)
    scores = cross_val_score(model, X, y, cv=skf, scoring='accuracy', n_jobs=1)
    
    print(f"  CV Scores: {scores}")
    print(f"  Mean CV Accuracy: {scores.mean():.4f} (+/- {scores.std() * 2:.4f})")
    
    return scores


def realtime_eeg_prediction(model, scaler, label_names, n_channels=4, sfreq=256):
    """
    Simulate real-time EEG prediction.
    """
    extractor = EEGFeatureExtractor(sfreq=sfreq, n_channels=n_channels)
    
    print(f"\n🧠 Real-time EEG Prediction Simulation")
    print(f"   Channels: {n_channels}")
    print(f"   Sample rate: {sfreq} Hz")
    print(f"   Press Ctrl+C to stop\n")
    
    # Your headset's channel names
    sim_channels = ['Fp1', 'Fp2', 'T7', 'T8'][:n_channels]
    
    try:
        while True:
            # Simulate 2 seconds of EEG data
            window_samples = int(2.0 * sfreq)
            fake_eeg = np.random.randn(window_samples, n_channels) * 50  # ~50 µV amplitude
            
            # Apply bandpass filter
            fake_eeg = bandpass_filter(fake_eeg, sfreq)
            
            # Extract features
            features, _ = extractor.extract_features(fake_eeg, sim_channels)
            features = features.reshape(1, -1)
            
            # Scale
            features_scaled = scaler.transform(features)
            
            # Predict
            pred_class = model.predict(features_scaled)[0]
            pred_label = label_names[pred_class]
            
            # Get probabilities if available
            if hasattr(model, 'predict_proba'):
                proba = model.predict_proba(features_scaled)[0]
                confidence = proba[pred_class]
                print(f"→ Predicted: {pred_label} (confidence: {confidence:.2%})")
            else:
                print(f"→ Predicted: {pred_label}")
            
            time.sleep(1.0)
            
    except KeyboardInterrupt:
        print("\n🛑 Stopped real-time simulation.")


# ============================================================
# MAIN EXECUTION
# ============================================================

if __name__ == "__main__":
    # ===========================================
    # CONFIGURATION - EDIT THESE VALUES DIRECTLY
    # ===========================================
    DATASET_PATH = "/project/ece601/EEG ALS/EEGET-ALS Dataset"
    N_CHANNELS = 4                    # Number of EEG channels (Fp1, Fp2, T7, T8)
    TEST_SIZE = 0.2                   # 20% for testing, 80% for training
    OUTPUT_FILE = "eeg_model_improved.joblib"
    RUN_SIMULATION = False            # Set to True to run real-time simulation after training
    # ===========================================
    
    print("=" * 60)
    print("EEG MOTOR IMAGERY CLASSIFICATION - IMPROVED VERSION")
    print("=" * 60)
    print(f"Dataset: {DATASET_PATH}")
    print(f"Channels: {N_CHANNELS} (Fp1, Fp2, T7, T8)")
    print(f"Test size: {TEST_SIZE}")
    
    # Prepare dataset
    train_data, test_data, scaler, label_names, feature_names = prepare_dataset(
        DATASET_PATH,
        n_channels=N_CHANNELS,
        test_size=TEST_SIZE,
        min_samples=10,
        simplify_labels=True
    )
    
    # Train and evaluate models
    results, best_model = train_and_evaluate_models(
        train_data, test_data, label_names, n_jobs=1
    )
    
    if best_model:
        name, model = best_model
        
        # Cross-validation on full scaled data
        X_full = np.vstack([train_data[0], test_data[0]])
        y_full = np.hstack([train_data[1], test_data[1]])
        
        # Clone model for CV
        from sklearn.base import clone
        cv_model = clone(model)
        cv_scores = cross_validate_best_model(X_full, y_full, cv_model, cv=5)
        
        # Save best model
        output_path = os.path.join(os.getcwd(), OUTPUT_FILE)
        joblib.dump({
            'model': model,
            'scaler': scaler,
            'labels': label_names,
            'feature_names': feature_names,
            'n_channels': N_CHANNELS,
            'cv_accuracy': cv_scores.mean(),
            'test_accuracy': results[name]['test_accuracy']
        }, output_path)
        print(f"\n✅ Model saved to: {output_path}")
        
        # Optional simulation
        if RUN_SIMULATION:
            realtime_eeg_prediction(model, scaler, label_names, n_channels=N_CHANNELS)
    
    print("\n" + "=" * 60)
    print("TRAINING COMPLETE")
    print("=" * 60)
```
